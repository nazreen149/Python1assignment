2.What is an interpreter in Python?


  - An **interpreter** is a program that executes code line-by-line directly, without needing to compile the entire program into machine language first.

  - In Python, this process works as follows:
1. **Bytecode Compilation:** When you run a Python script, the source code is first compiled into intermediate code called **bytecode** (stored as `.pyc` files).
2. **Virtual Machine (PVM):** The Python Virtual Machine (PVM) then interprets this bytecode line-by-line and translates it into machine-level instructions that your computer's CPU can understand and execute.




3.What are pre-defined keywords in Python?

 Key Characteristics:

4.Can keywords be used as variable names?
  - **No**, keywords cannot be used as variable names, function names, or any other identifiers in Python because they are reserved words with pre-defined meanings in the language's grammar.

5.What is mutability in Python?
    

   - **Mutability** refers to the ability to modify or change an object's state or content after it has been created.

 -  Mutable vs. Immutable Objects:
* **Mutable Objects:** Can be changed after creation (e.g., `list`, `dict`, `set`)
* **Immutable Objects:** Cannot be changed after creation (e.g., `int`, `float`, `string`, `tuple`)

6.Why are lists mutable, but tuples are immutable?
  - **Lists** are designed for dynamic data collections where elements need to be added, removed, or changed. They are stored in over-allocated, dynamic memory arrays, allowing in-place modifications.
  - **Tuples** are designed for fixed, heterogeneous data collections (like a database record). They are stored directly in static, fixed-size memory blocks, making them memory-efficient, faster to access, and safe from accidental modifications.

7.What is the difference between “==” and “is” operators in Python?
  - **(Value Equality):** Compares the **values** or contents of two objects to see if they are equal.
  - **(Reference Identity):** Checks if two variables point to the exact **same object in memory** (i.e., they share the same memory address, which can be verified using `id()`).

8.What are logical operators in Python?
  - Logical operators are used to combine conditional statements. Python has three logical operators:
    * **`and`**: Returns `True` if both statements are true (e.g., `x and y`).
    * **`or`**: Returns `True` if at least one of the statements is true (e.g., `x or y`).
    * **`not`**: Reverses the logical state; returns `True` if the statement is false, and vice versa (e.g., `not x`).

9.What is type casting in Python?
  - **Type casting** (or type conversion) is the process of converting a variable from one data type to another. In Python, this can happen in two ways:
    * **Implicit Type Conversion:** Python automatically converts one data type to another to prevent data loss (e.g., adding an integer to a float automatically results in a float).
    * **Explicit Type Conversion (Type Casting):** The programmer manually converts a data type using built-in functions such as `int()`, `float()`, `str()`, `list()`, `tuple()`, etc.

10.What is the difference between implicit and explicit type casting?
  - **Implicit Type Casting (Coercion):**
    * Done automatically by the Python interpreter.
    * No user intervention or explicit functions are required.
  - **Explicit Type Casting (Casting):**
    * Done manually by the developer.
    * Requires built-in casting functions `float()`, `str()'

11.What is the purpose of conditional statements in Python?
  - **Conditional statements** (like `if`, `elif`, and `else`) allow a program to make decisions and execute different blocks of code depending on whether a given condition evaluates to `True` or `False`.
  - They control the decision-making flow of the program, making the code dynamic and responsive to various inputs or states.

12.How does the elif statement work?
  - **`elif`** is short for "else if". It is used to check multiple conditions sequentially.
  - It only executes if all previous `if` and `elif` conditions in the chain evaluated to `False`.
  - Once Python finds a condition that evaluates to `True`, it executes that specific block and completely skips evaluating any remaining `elif` or `else` statements in the structure.

13.What is the difference between for and while loops?
  - **`for` loop:** Used for definite iteration when you know the number of iterations beforehand (e.g., looping over a range or a list).
  - **`while` loop:** Used for indefinite iteration to repeat code as long as a condition remains true.

14.Describe a scenario where a while loop is more suitable than a for loop?
  - A **`while` loop** is best when you don't know ahead of time how many times you need to loop, and instead want to keep going until a certain condition changes.
  - **Scenario:** Getting input from a user. For instance, if you are asking a user to type their password or enter `exit` to close a program, you can't predict how many attempts it will take them. A `while` loop is perfect here because it will repeatedly ask for input and only stop once the correct action is taken.

### Q13. Write a Python program to find the sum of all even numbers between 1 and 50.

### Q14. Write a program to reverse a string using a while loop.

```python
string = input("Enter a string: ")
reversed_string = ""
i = len(string) - 1

while i >= 0:
    reversed_string += string[i]
    i -= 1

print("Reversed string:", reversed_string)
```

### Q15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.

```python
# Get input from the user
num = int(input("Enter a number to calculate its factorial: "))

# Initialize variables
factorial = 1
i = 1

# Calculate factorial using a while loop
while i <= num:
    factorial *= i
    i += 1

print(f"The factorial of {num} is: {factorial}")
```
